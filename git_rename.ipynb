# Rename *.markdown files to *.md files with git

```python
from pathlib import Path
import subprocess
```

```python
base_path = Path(".")
base_path
```

```python
pwd
```

```python
for p in base_path.glob("**/*.markdown"):
    #print(p)
    p_md = p.parent.joinpath(p.stem + ".md")
    subprocess.run(["git", "mv", str(p), str(p_md)])
```

```python
for p in base_path.glob("**/*.md"):
    print(p)
```
